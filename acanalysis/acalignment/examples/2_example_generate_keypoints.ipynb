```python
import json
from pathlib import Path
```

```python
from acanalysis.acalignment.generate_keypoints import generate_keypoint_file,ori_lookup
from utils import get_swc_path_from_roi,get_roi_label
```

```python
# Alignment json describing the set of segmented ROI's (SROI's) from each
# section image to be jointly aligned, the image (zarr) data cutout regions
# underlying the segmentation, and file path information to locate
# the zarr datasets and skeletons
with open("align_S32_S33_roi1_mip0.json") as f:
    alignjs = json.load(f)
outpath = Path(alignjs["output_path"])
is_tar = False
swap_xyz = []
```

```python
# detect keypoints (endpoints of skeletons near the surface) for each SROI
for s in alignjs["sections"]:
    ori = s["ori"]
    oristr = ori_lookup(ori)[1]
    surfsup = (oristr == "POS")
    #co_coords = s["cutouts"][0]
    #roi_coords = [[],[],[co_coords["z"][0],co_coords["z"][1]]]
    roi_coords = None
    for i,tid in enumerate(s["tiles"]):
        roi_id = get_roi_label(s,i)
        swcpath = Path(s["swc_path"]) / Path(tid + ".swc")
        #swcpath = Path(s["swc_path"]) / Path(s["section_id"] + "_" + s["acq_id"]) / Path(tid + ".swc")
        surfmap = outpath / Path(roi_id + "_" + oristr + ".npy")
        outputpath = outpath / Path(roi_id + "_keypoints.json")
        generate_keypoint_file(swcpath=swcpath,
                               outputpath=outputpath,
                               is_tar=is_tar,
                               swcmip=s["swc_mip"],
                               ori=ori,
                               swap_xyz=swap_xyz,
                               tile_name=roi_id+"_",
                               surf_file=surfmap,
                               roi_coords=roi_coords,
                               mincablelength=50,#50,
                               minradius=None,#2,
                               distance=100)
        print("saved keypoints to " + str(outputpath))
```

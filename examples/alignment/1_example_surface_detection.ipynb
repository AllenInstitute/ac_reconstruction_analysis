```python
import numpy
import json
from pathlib import Path
```

```python
from acpreprocessing.surface_detection.surface_detection import detect_surface
from acanalysis.acalignment.utils.kevin_utils import surf_inputs_from_alignjs
```

```python
# Alignment json describing the set of segmented ROI's (SROI's) from each
# section image to be jointly aligned, the image (zarr) data cutout regions
# underlying the segmentation, and file path information to locate
# the zarr datasets and skeletons
align_json = "align_S32_S33_roi2_mip1_240403.json"
with open(align_json) as f:
    alignjs = json.load(f)
outpath = Path(alignjs["output_path"])
if not outpath.exists():
    outpath.mkdir(parents=True)
```

```python
# generate a surface map from downsampled (mip level 2) data for each SROI
miplvl = 2
inputs,outputs = surf_inputs_from_alignjs(alignjs,miplvl)
```

```python
print(inputs)
```

```python
for kwargs,outputpath in zip(inputs,outputs):
    print("saving output to " + str(outputpath))
    G = numpy.array(detect_surface(**kwargs))
    numpy.save(outputpath,G)
```
